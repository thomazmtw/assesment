```python
# Q12. Filtering & Sorting
print("Final Score > 75:\n", df[df['Final_Score'] > 75])
print("Attendance < 75:\n", df[df['Attendance'] < 75])
print("Data Science > 80:\n", df[(df['Department'] == 'Data Science') & (df['Final_Score'] > 80)])
print("Sorted by Final Score:\n", df.sort_values('Final_Score', ascending=False))
```

**Q 13**

```python
# Q13. Feature Creation
df['Average_Technical_Score'] = df[['Python_Score', 'SQL_Score', 'Pandas_Score', 'NumPy_Score']].mean(axis=1)
print(df)
conditions = [
    df['Average_Technical_Score'] >= 90,
    df['Average_Technical_Score'] >= 75,
    df['Average_Technical_Score'] >= 60
]
choices = ['Excellent', 'Good', 'Average']
df['Performance'] = np.select(conditions, choices, default='Needs Improvement')
print(df)
```

**Q 14**

```python
# Q14. GroupBy Analysis
dept_summary = df.groupby('Department').agg(
    Student_Count=('Student_ID', 'count'),
    Avg_Final_Score=('Final_Score', 'mean'),
    Max_Final_Score=('Final_Score', 'max'),
    Min_Final_Score=('Final_Score', 'min')
).sort_values('Avg_Final_Score', ascending=False)
print(dept_summary)
```

**Q 15**

```python
# Q15. NumPy & Pandas Integration
scores = df['Final_Score'].dropna().values

print("Mean:", scores.mean())
print("Median:", np.median(scores))
print("Std:", scores.std())
print("25th Percentile:", np.percentile(scores, 25))
print("75th Percentile:", np.percentile(scores, 75))

print(df[df['Final_Score'] > scores.mean()])
```

**Q 16**

```python
#Q16. Student Performance Analysis
df = pd.read_csv('/content/drive/MyDrive/dsa /asessment dataset/student_performance.csv')
```

```python
#A - data cleaning

print("Missing values:\n", df.isnull().sum())
df = df.fillna(df.mean(numeric_only=True))
```

```python
# B - feature engineering

df['Average_Score'] = df[['Python_Score', 'SQL_Score', 'Pandas_Score', 'NumPy_Score']].mean(axis=1)
df['Pass_Status'] = np.where(df['Average_Score'] >= 50, 'Pass', 'Fail')
print(df.head)
```

```python
# C - analysis

print("Top 5 Students:\n", df.nlargest(5, 'Average_Score'))
print("Top Department:", df.groupby('Department')['Average_Score'].mean().idxmax())
print("Avg Attendance by Dept:\n", df.groupby('Department')['Attendance'].mean())
print("Pass/Fail Count:\n", df['Pass_Status'].value_counts())
```

```python
# D - summary
summary = df.groupby('Department').agg(
    Total_Students=('Student_ID', 'count'),
    Average_Score=('Average_Score', 'mean'),
    Average_Attendance=('Attendance', 'mean'),
    Pass_Count=('Pass_Status', lambda x: (x == 'Pass').sum()),
    Fail_Count=('Pass_Status', lambda x: (x == 'Fail').sum())
)
print(summary)
```

**Q 18**

```python
df = pd.read_csv('/content/drive/MyDrive/dsa /asessment dataset/employee.csv')
df['Salary'] = pd.to_numeric(df['Salary'], errors='coerce')
df['Experience'] = pd.to_numeric(df['Experience'], errors='coerce')
print(df)
```

```python
# A - NumPy Analysis

salary = df['Salary'].dropna().values
perf = df['Performance_Score'].dropna().values

print("Mean Salary:", salary.mean())
print("Median Salary:", np.median(salary))
print("Std Salary:", salary.std())
print("Mean Performance:", perf.mean())
```

```python
# B - Filtering

filtered = df[(df['Experience'] > 3) & (df['Performance_Score'] >= 80)]
print(filtered)
```

```python
# C - Department Analysis

dept_summary = df.groupby('Department').agg(
    Average_Salary=('Salary', 'mean'),
    Average_Performance=('Performance_Score', 'mean'),
    Total_Projects=('Projects_Completed', 'sum'),
    Employee_Count=('Employee_ID', 'count')
)
print(dept_summary)
print("Highest Performance Dept:", dept_summary['Average_Performance'].idxmax())
```

```python
# D - Analytical Interpretation

flagged = df[(df['Salary'] > salary.mean()) & (df['Performance_Score'] < perf.mean())]
print(flagged[['Employee_ID', 'Department', 'Salary', 'Performance_Score']])
```
