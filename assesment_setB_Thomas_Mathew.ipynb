```python
#parking fee calculator
vehicle_type = input("Enter vehicle type (car/bike/suv): ")
hours = int(input("Enter parking duration in hours: "))

rates = {
    "car": (50, 20),
    "bike": (30, 10),
    "suv": (70, 30)
}

if vehicle_type not in rates:
    print("Invalid vehicle type")
elif hours <= 0:
    print("Invalid parking duration")
else:
    first_rate, add_rate = rates[vehicle_type]
    fee = first_rate if hours <= 2 else first_rate + (hours - 2) * add_rate

    print(f"Vehicle: {vehicle_type}")
    print(f"Hours: {hours}")
    print(f"Parking Fee: {fee}")
```

```python
#Movie Theatre Seat Booking Analyzer
n = int(input("Enter number of shows: "))
total_booked = 0
total_available = 0
high_demand_count = 0
highest_occ = -1
highest_show = ""
for _ in range(n):
    show = input("Enter Show Name: ")
    total = int(input("Enter Total Seats: "))
    booked = int(input("Enter Booked Seats: "))

    available = total - booked
    occ = (booked / total) * 100

    if occ >= 80:
        category = "House Full"
    elif occ >= 60:
        category = "High Demand"
    elif occ >= 40:
        category = "Moderate"
    else:
        category = "Low Demand"

    if occ >= 60:
        high_demand_count += 1

    if occ > highest_occ:
        highest_occ = occ
        highest_show = show
    total_booked += booked
    total_available += available

    print(f"{show} - Available: {available} - Occupancy: {occ:.2f}% - {category}")

print(f"Highest Occupancy: {highest_show}")
print(f"Total Booked Seats: {total_booked}")
print(f"Total Available Seats: {total_available}")
print(f"High Demand/House Full Shows: {high_demand_count}")
```

```python
#Employee Leave Management System
class Employee:
    def __init__(self, emp_id, name, dept, balance):
        self.employee_id = emp_id
        self.employee_name = name
        self.department = dept
        self.leave_balance = balance

    def apply_leave(self, days):
        if days <= 0:
            print("Invalid Leave Request")
        elif days > self.leave_balance:
            print("Insufficient Leave Balance")
        else:
            self.leave_balance -= days
            print("Leave Approved")

    def cancel_leave(self, days):
        if days <= 0:
            print("Invalid Cancellation")
        else:
            self.leave_balance += days

    def check_balance(self):
        print(f"Leave Balance: {self.leave_balance}")

    def display_employee(self):
        print(f"Employee ID: {self.employee_id}")
        print(f"Employee Name: {self.employee_name}")
        print(f"Department: {self.department}")
        print(f"Leave Balance: {self.leave_balance}")


emp_info = input().split()
emp_id = emp_info[0]
emp_name = emp_info[1]
dept = emp_info[2]
init_balance = int(emp_info[3])

emp = Employee(emp_id, emp_name, dept, init_balance)

n = int(input())

for _ in range(n):
    parts = input().split()
    command = parts[0]

    if command == "APPLY":
        emp.apply_leave(int(parts[1]))
    elif command == "CANCEL":
        emp.cancel_leave(int(parts[1]))
    elif command == "BALANCE":
        emp.check_balance()
    elif command == "DISPLAY":
        emp.display_employee()
```
